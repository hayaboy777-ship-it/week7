```python
# ── [실습 7] 공정 WIDE 불러오기와 기초 확인 ─────────────────────────────────
wide = pd.read_csv(HW)
print(wide.shape, int(wide.isna().sum().sum()))  # (120, 8) 25
wide2 = wide.fillna(wide.mean(numeric_only=True))
print(int(wide2.isna().sum().sum()))  # 0
```

```python
# ── [실습 8] 시각화 리포트 완성과 저장 — 캡스톤 (2×2) ───────────────────────
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.hist(wide2["소입로온도1존"], bins=15)
plt.title("소입로온도1존 분포")

plt.subplot(2, 2, 2)
vc = wide2["라인"].value_counts()
plt.bar(vc.index, vc.values)
plt.title("라인별 개수")

plt.subplot(2, 2, 3)
plt.scatter(wide2["소입로온도1존"], wide2["OP값"], alpha=0.5)
plt.title("소입로온도1존 vs OP값")

plt.subplot(2, 2, 4)
rc = wide2["result"].value_counts()
plt.bar(rc.index, rc.values)
plt.title("판정 결과")
plt.tight_layout()
plt.show()
# print(dict(wide2["result"].value_counts()))  # {'정상': 109, '이상': 11}
# print("=== 17_02 전체 실행 완료 ===")
```
