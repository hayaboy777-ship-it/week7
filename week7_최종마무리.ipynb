```python
sns.set_theme(style="whitegrid", font='Malgun Gothic')
```

```python
HT = "..//data//18_열처리.csv"
```

```python
df = pd.read_csv(HT)
```

```python
# ── [개념] scatterplot — 두 변수 관계 ──────────────────────────────────────
# 형식: sns.scatterplot(data=df, x=, y=). alpha로 반투명, hue=범주로 그룹 색 구분.
```

```python
# 9시 20분
```

```python
# ── [실습 1] 산점도로 두 변수 관계 보기 (소입로온도-세정기) ─────────────────
sns.scatterplot(data=df, x="소입로온도", y="세정기", alpha=0.5)
plt.title("소입로온도와 세정기의 관계")
plt.xlabel("소입로온도(℃)")
plt.ylabel("세정기")
plt.show()
```

```python
df['소입로온도'].corr(df['세정기'])
```

```python
# ── [실습 2] 판정별 산점도 패턴 (소입로온도-제어출력 hue) ───────────────────
sns.scatterplot(data=df, x="소입로온도", y="제어출력", hue="판정")
plt.title("판정별 소입로온도-제어출력 관계")
plt.show()
```

```python
df['소입로온도'].corr(df['제어출력'])
```

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("센서 상관관계")
plt.show()
```

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("센서 상관관계")
plt.show()
```

```python
for a, b in [
    ("소입로온도", "제어출력"),
    ("소입로온도", "건조출력"),
    ("소입로온도", "세정기"),
]:
    print(f"{a}-{b}: {df[a].corr(df[b]):.2f}")  # -0.74 / -0.01 / 0.03
```
