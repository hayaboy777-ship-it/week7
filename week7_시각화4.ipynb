```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
HT = "..//data//18_열처리.csv"
```

```python
df = pd.read_csv(HT)
```

```python
sns.set_theme(style="whitegrid", font='Malgun Gothic')
# plt.rcParams['font.family'] = "Malgun Gothic"
sns.histplot(df, x='소입로온도', hue="판정",bins=17, kde=True)
plt.show()
```

```python
df.head()
```

```python
# ── [실습 1] 첫 분포 그래프 그리기 (건조출력) ───────────────────1시 32분
```

```python
df['건조출력']
```

```python
sns.histplot(data=df, x="건조출력")
plt.title("건조출력 분포")
plt.xlabel("건조출력(%)")
plt.ylabel("개수")
plt.show()
```

```python
# ── [실습 2] bins·kde 조절 (건조출력) ───────────────────────────────────────
```

```python
# 1시 36분까지
```

```python
# ── [실습 2] bins·kde 조절 (건조출력) ───────────────────────────────────────
for b in [10, 20, 40]:
    sns.histplot(data=df, x="건조출력", bins=b, kde=True)
    plt.title(f"건조출력 분포 bins={b}")
    plt.show()
```

```python
# ── [실습 3] 정상·이상 분포 비교 (소입로온도, hue) ──────────────────────────
```

```python
sns.histplot(data=df, x="소입로온도", hue="판정", kde=True)
plt.title("판정별 소입로온도 분포")
plt.xlabel("소입로온도(℃)")
plt.show()
```

```python
# ("소입로온도", "℃"), ("제어출력", "%"), ("건조출력", "%") 분포 한눈에 보기
```

```python
# 1시 50분까지
```

```python
# ── [실습 4] 여러 센서 분포 한눈에 보기 ─────────────────────────────────────
for col, unit in [("소입로온도", "℃"), ("제어출력", "%"), ("건조출력", "%")]:
    sns.histplot(data=df, x=col, bins=20, kde=True)
    plt.title(f"{col} 분포")
    plt.xlabel(f"{col}({unit})")
    plt.show()
```

```python
# ─ [실습 6] 범주별 박스플롯 비교 (소입로온도, 라인·교차) 라인별  ───────────────────
```

```python
df['라인'].value_counts()
```

```python
sns.boxplot(data=df, x="라인", y="소입로온도")  # 라인별
plt.title("라인별 소입로온도 분포")
plt.show()
```

```python
sns.boxplot(data=df, x="라인", y="소입로온도", hue="판정")  # 라인×판정 교차
plt.title("라인별·판정별 소입로온도 비교")
plt.show()
```

```python
# ── [실습 7] countplot·barplot (라인 개수·평균 제어출력) ────────────────────
```

```python
df.head()
```

```python
sns.boxplot(data=df, x="라인", y="소입로온도")  # 라인별
plt.title("라인별 소입로온도 분포")
plt.show()
```

```python
sns.barplot(data=df, x="라인", y='소입로온도')  # 라인별
plt.title("라인별 데이터 개수")
plt.show()
```

```python
sns.barplot(data=df, x="라인", y='소입로온도', palette='pastel', legend=True)  # 라인별
plt.title("라인별 데이터 개수")
plt.show()
```

```python
sns.countplot(data=df, x="라인")  # 라인별
plt.title("라인별 데이터 개수")
plt.show()
```

```python
sns.countplot(data=df, x="라인", hue='판정')  # 라인별
plt.title("라인별 데이터 개수")
plt.show()
```

```python
# ── [실습 8] 라인별 이상 비율 비교 ───
```

```python
sns.countplot(data=df, x="라인", hue="판정")
plt.title("라인별 정상·이상 개수")
plt.show()
```
